 Simple Sum Calculator:
Keep asking the user for numbers and calculate the sum until they enter 0.

```python
total = 0

while True:
    num = int(input("Enter a number: "))

    if num == 0:
        break

    total += num

print("Sum:", total)
```

Input a number and reverse its digits using a while loop

```python
num = int(input("Enter a number: "))
reverse = 0
while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10
print("Reversed number:", reverse)
```

Generate the first N terms of the Fibonacci sequence using a while loop.

```python
N = int(input("Enter N: "))
a = 0
b = 1
count = 0
while count < N:
    print(a, end=" ")
    a, b = b, a + b
    count += 1
```

Write a program to check whether a number is a palindrome (e.g., 121 is a palindrome).


```python
num = int(input("Enter a number: "))
original = num
reverse = 0
while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10
if original == reverse:
    print("Palindrome")
else:
    print("Not a palindrome")
```

Print numbers from 1 to 50. For multiples of 3, print "Fizz", for multiples of 5 print "Buzz", and
for multiples of both, print "FizzBuzz".

```python
for i in range(1, 51):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```
